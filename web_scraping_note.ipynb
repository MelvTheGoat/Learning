```python
from bs4 import BeautifulSoup as soup
import requests
```

```python
data = requests.get("https://books.toscrape.com/")
data = data.text
data = soup(data, 'html.parser')
# data
```

```python
container = data.find_all("ol", {"class":"row"})    #find, find_all
len(container)
```

```python
name = data.find("h3").a["title"]
name
```

```python
price = data.find("p", {"class":"price_color"})
price.text
```

```python
book_name = []
book_price = []
    
for i in container:
    na = i.find_all("h3")
    for j in na:
        book_name.append(j.a.text)

    price = i.find_all("p", {"class":"price_color"})
    for x in price:
        pr = x.text
        book_price.append(pr)
                       
                       
#     for j in i:
#         book_name.append(j.find("h3").a["title"])
print(book_name)
print(book_price)
```

```python
import pandas as pd

dataset = pd.DataFrame({"Name of book": book_name, "Price":book_price})
dataset
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()))
driver.get("https://www.jumia.com.ng/televisions/")
source = driver.page_source
driver.quit()
```

Service helps Python start and manage the ChromeDriver program. like the translator between Python and Chrome.
    
ChromeDriver needs to be compatible with your installed Chrome browser.
    
Instead of manually searching online for the correct ChromeDriver, ChromeDriverManager helps download and manage the appropriate driver.
    

```python
data = soup(source, 'html.parser')
data.text
```

```python
print(data.prettify())
```

```python
container = data.find_all("div", {"class":"-phm -pvxs row _no-g _4cl-3cm-shs"})
len(container)
```

```python
name = data.find("h3", {"class":"name"})
name.text
```

```python
price = data.find("div", {"class":"prc"})
price.text
```

```python
product_name =[]
product_price = []

for i in container:
    name = i.find_all("h3", {"class":"name"})
    for j in name:
        product_name.append(j.text)

    price = i.find_all("div", {"class":"prc"})
    for j in price:
        product_price.append(j.text)
product_price
```

```python
import pandas as pd
```

```python
dataset = pd.DataFrame({"product":product_name, "price":product_price})
dataset
```
